# Copy 'Unknown' Class Images

This script copies the `Unknown` class images (which contain non-oral images from Kaggle) from the old 5-class dataset to the newly created `patientwise` 5-class dataset.
Since these images do not belong to the Sri Lankan dataset, there is no risk of patient-level data leakage when blindly copying the previously established train/val/test splits.

```python
import os
import shutil

SOURCE_DATASET = "/Users/alokdwivedi/dev/oral_cavity_5class_dataset"
TARGET_DATASET = "/Users/alokdwivedi/dev/oral_cavity_5class_dataset_patientwise"

SPLITS = ['train', 'val', 'test']

for split in SPLITS:
    src_unknown_dir = os.path.join(SOURCE_DATASET, split, 'Unknown')
    tgt_unknown_dir = os.path.join(TARGET_DATASET, split, 'Unknown')
    
    if os.path.exists(src_unknown_dir):
        print(f"Copying {src_unknown_dir} -> {tgt_unknown_dir}")
        if os.path.exists(tgt_unknown_dir):
            shutil.rmtree(tgt_unknown_dir)
            
        shutil.copytree(src_unknown_dir, tgt_unknown_dir)
        print(f"Successfully copied {len(os.listdir(tgt_unknown_dir))} images to {split}/Unknown")
    else:
        print(f"WARNING: Source directory {src_unknown_dir} does not exist!")

print("Finished copying Unknown class images.")
```
